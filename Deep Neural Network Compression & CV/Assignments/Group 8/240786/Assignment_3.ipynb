````python
import numpy as np
import cv2
from matplotlib import pyplot as plt


# CONVULUTION
def convolve(img, kernel):
    # flip the kernel for convolution
    kernel = np.flipud(np.fliplr(kernel))

    kh, kw = kernel.shape
    ih, iw = img.shape[:2]

    pad_h = kh // 2
    pad_w = kw // 2

    # reflect padding
    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    output = np.zeros_like(img, dtype=float)

    # manual convolution
    for i in range(ih):
        for j in range(iw):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)

    return output


# AVERAGE BLUR

def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=float) / (ksize * ksize)
    return convolve(img, kernel)

# print the box kernel for average blur
ksize = 5
box_kernel = np.ones((ksize, ksize), dtype=float) / (ksize * ksize)
print("Box Kernel (Average Blur):")
print(box_kernel)



def average_blur_rgb(img, ksize):
    out = np.zeros_like(img, dtype=float)
    kernel = np.ones((ksize, ksize), dtype=float) / (ksize * ksize)

    # process each channel separately
    for c in range(3):
        out[:, :, c] = convolve(img[:, :, c], kernel)

    return out


# GAUSSIAN BLUR

def gaussian_kernel(ksize, sigma):
    k = ksize // 2
    x, y = np.meshgrid(np.arange(-k, k+1), np.arange(-k, k+1))

    # gaussian formula
    g = np.exp(-(x**2 + y**2) / (2 * sigma**2))
    g = g / np.sum(g)    # normalise

    return g

# printing gaussian kernel
gk = gaussian_kernel(7, 1.5)
print("Gaussian Kernel:")
print(gk)



def gaussian_blur_gray(img, ksize, sigma):
    gk = gaussian_kernel(ksize, sigma)
    return convolve(img, gk)


def gaussian_blur_rgb(img, ksize, sigma):
    gk = gaussian_kernel(ksize, sigma)
    out = np.zeros_like(img, dtype=float)

    for c in range(3):
        out[:, :, c] = convolve(img[:, :, c], gk)

    return out



img = cv2.imread("img_4.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# average blur results
avg_gray = average_blur_gray(gray, 5)
avg_rgb = average_blur_rgb(img, 5)

# Gaussian blur results
gauss_gray = gaussian_blur_gray(gray, 7, 1.5)
gauss_rgb = gaussian_blur_rgb(img, 7, 1.5)

# displaying results

plt.figure(figsize=(12, 10))

plt.subplot(2, 3, 1)
plt.title("Original RGB")
plt.imshow(img)
plt.axis("off")

plt.subplot(2, 3, 2)
plt.title("Average Blur (RGB)")
plt.imshow(avg_rgb.astype(np.uint8))
plt.axis("off")

plt.subplot(2, 3, 3)
plt.title("Gaussian Blur (RGB)")
plt.imshow(gauss_rgb.astype(np.uint8))
plt.axis("off")

plt.subplot(2, 3, 4)
plt.title("Original Gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 5)
plt.title("Average Blur (Gray)")
plt.imshow(avg_gray, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 6)
plt.title("Gaussian Blur (Gray)")
plt.imshow(gauss_gray, cmap='gray')
plt.axis("off")

plt.show()
````

#### Task - Sobel Edge Detection (NumPy Only)
Implement Sobel X and Sobel Y filters manually using convolution. Apply on the grayscale image.

Compute:
1. Gx
2. Gy
3. Gradient magnitude

Normalize to 0–255 and save the edge map. Threshold if you wish to, if you do then make the thresholding part clear using code comments.

````python


# Sobel kernels
sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=float)

sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=float)

print("Sobel X Kernel:")
print(sobel_x)
print("\nSobel Y Kernel:")
print(sobel_y)

# apply Sobel on grayscale image
Gx = convolve(gray, sobel_x)
Gy = convolve(gray, sobel_y)

# gradient magnitude
grad_mag = np.sqrt(Gx**2 + Gy**2)

# normalize to 0–255
grad_norm = (grad_mag / grad_mag.max()) * 255
grad_norm = grad_norm.astype(np.uint8)

# thresholding
threshold = 80
edge_map = grad_norm.copy()
edge_map[edge_map < threshold] = 0     # if it is below this threshold then it's not an edge
edge_map[edge_map >= threshold] = 255  # and if it's above that threshold it's and edge

# displaying results

plt.figure(figsize=(14, 10))

plt.subplot(2, 3, 1)
plt.title("Grayscale Image")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 2)
plt.title("Sobel Gx")
plt.imshow(np.abs(Gx), cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 3)
plt.title("Sobel Gy")
plt.imshow(np.abs(Gy), cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 4)
plt.title("Gradient Magnitude (0–255)")
plt.imshow(grad_norm, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 5)
plt.title("Edge Map (Thresholded)")
plt.imshow(edge_map, cmap='gray')
plt.axis("off")

plt.tight_layout()
plt.show()
````

#### Task - Frequency Analysis (LPF & HPF Verification)
Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python

def magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)

    # magnitude spectrum
    mag = np.log(1 + np.abs(fshift))
    return mag


# original maginitude spectrum
mag_original = magnitude_spectrum(gray)

# gaussian blur
sigma = 1.5
ksize = 7
blurred = gaussian_blur_gray(gray, ksize, sigma)

mag_blur = magnitude_spectrum(blurred)

# unsharped mask
alpha = 1.5
mask = gray - blurred
sharp = gray + alpha * mask
sharp = np.clip(sharp, 0, 255).astype(np.uint8)

mag_sharp = magnitude_spectrum(sharp)


# displaying results

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.title("Original Gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 2)
plt.title("Gaussian Blurred")
plt.imshow(blurred, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 3)
plt.title(f"Unsharp Mask (alpha={alpha})")
plt.imshow(sharp, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 4)
plt.title("Magnitude Spectrum (Original)")
plt.imshow(mag_original, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 5)
plt.title("Magnitude Spectrum (Blurred — LPF)")
plt.imshow(mag_blur, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 6)
plt.title("Magnitude Spectrum (Sharpened — HPF)")
plt.imshow(mag_sharp, cmap='gray')
plt.axis("off")

plt.tight_layout()
plt.show()
````

````python


R = img_part5[:, :, 0].astype(float)
G = img_part5[:, :, 1].astype(float)
B = img_part5[:, :, 2].astype(float)

# yellow pixels , R high and G high , B low
yellow_score = R + G - 2 * B

# clip to valid range
yellow_score = np.clip(yellow_score, 0, 255)

yellow_smooth = gaussian_blur_gray(yellow_score, ksize=5, sigma=1.0)

sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=float)

sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=float)

gx = convolve(yellow_smooth, sobel_x)
gy = convolve(yellow_smooth, sobel_y)

edge_mag = np.sqrt(gx**2 + gy**2)

edge_norm = (edge_mag / edge_mag.max()) * 255
edge_norm = edge_norm.astype(np.uint8)

threshold = 80
yellow_edges = edge_norm.copy()
yellow_edges[yellow_edges < threshold] = 0
yellow_edges[yellow_edges >= threshold] = 255


# displaying results
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.title("Original Image")
plt.imshow(img_part5)
plt.axis("off")

plt.subplot(2, 3, 2)
plt.title("Yellow Likelihood (R + G - 2B)")
plt.imshow(yellow_score, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 3)
plt.title("Smoothed Yellow Map")
plt.imshow(yellow_smooth, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 4)
plt.title("Sobel Edge Magnitude")
plt.imshow(edge_norm, cmap='gray')
plt.axis("off")

plt.subplot(2, 3, 5)
plt.title("Final Yellow Leaf Edge Map")
plt.imshow(yellow_edges, cmap='gray')
plt.axis("off")

plt.tight_layout()
plt.show()

````
