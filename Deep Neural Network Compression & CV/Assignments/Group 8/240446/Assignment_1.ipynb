````python
#Convert to float for math
gray_float=gray.astype(np.float32)
#FFT
f=np.fft.fft2(gray_float)
#shift low frequencies to center
fshift=np.fft.fftshift(f)
````

````python
#Magnitude spectrum
magnitude=20*np.log(np.abs(fshift)+1) #multiplying by 20 just makes the spectrum look clearer
#phase spectrum
phase=np.angle(fshift)
#display magnitude spectrum
plt.figure(figsize=(6,6))
plt.imshow(magnitude,cmap='gray')
plt.title("Magnitude Spectrum")
plt.axis('off')
plt.show()
````

````python
#LPF
rows,cols= gray.shape
crow,ccol=rows//2,cols//2

#radius 10
radius1=10
mask_lpf1=np.zeros((rows,cols),np.uint8) #matrix full of zeros same size as that of gray
cv2.circle(mask_lpf1,(ccol,crow),radius1,1,-1) #matrix having a centre circle of 1s other points 0
f_lpf1=fshift*mask_lpf1 #applying the mask
ishift_lpf1=np.fft.ifftshift(f_lpf1)
img_back_lpf1=np.fft.ifft2(ishift_lpf1)
img_back_lpf1=np.abs(img_back_lpf1)
plt.figure(figsize=(6,6))
plt.imshow(img_back_lpf1,cmap='gray')
plt.title("LPF with radius 10")
plt.axis('off')
plt.show()

#radius 30
radius2=30
mask_lpf2=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf2,(ccol,crow),radius2,1,-1)
f_lpf2=fshift*mask_lpf2
ishift_lpf2=np.fft.ifftshift(f_lpf2)
img_back_lpf2=np.fft.ifft2(ishift_lpf2)
img_back_lpf2=np.abs(img_back_lpf2)
plt.figure(figsize=(6,6))
plt.imshow(img_back_lpf2,cmap='gray')
plt.title("LPF with radius 30")
plt.axis('off')
plt.show()

#radius=80
radius3=80
mask_lpf3=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf3,(ccol,crow),radius3,1,-1)
f_lpf3=fshift*mask_lpf3
ishift_lpf3=np.fft.ifftshift(f_lpf3)
img_back_lpf3=np.fft.ifft2(ishift_lpf3)
img_back_lpf3=np.abs(img_back_lpf3)
plt.figure(figsize=(6,6))
plt.imshow(img_back_lpf3,cmap='gray')
plt.title("LPF with radius 80")
plt.axis('off')
plt.show()
````
