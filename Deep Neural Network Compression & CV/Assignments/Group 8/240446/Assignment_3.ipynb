````python
import cv2
#converting PIL image to an RGB numpy image
img_rgb=np.array(img)
#converting image to grey scale
gray=cv2.cvtColor(img_rgb,cv2.COLOR_RGB2GRAY)
````

````python
#convolution function (greyscale)
import numpy as np
import matplotlib.pyplot as plt

def convolve(img, kernel):
    k = kernel.shape[0]
    pad = k // 2

    # padding-without padding border pixels have no neighbours which leads to poor blur at the borders
    padded = np.pad(img, pad_width=pad, mode='reflect') #np.pad adds pixels around image borders

    H, W = img.shape
    out = np.zeros((H, W), dtype=np.float64)

    # creating a sliding window that moves pixel by pixel
    for i in range(H):
        for j in range(W):
            region = padded[i:i+k, j:j+k]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
#average blur
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float64)
    kernel=kernel/kernel.sum()  # to get sum as 1
    return convolve(img, kernel)
````

````python
#testing our average blur function
blurred_gray = average_blur_gray(gray, ksize=11)
plt.figure(figsize=(6,6))
plt.imshow(blurred_gray, cmap='gray')
plt.title('Average blur on greyscale')
plt.axis('off')
plt.show()
#as we increase ksize blurness increases(more neighbouring pixels can influence the resultant value)
````

````python
#Repeating for rgb
def average_blur_rgb(img, ksize):
    out = np.zeros_like(img, dtype=np.float64)

    #applying average_blur_gray separately for R,G and B channels
    for c in range(3):
        out[..., c] = average_blur_gray(img[..., c], ksize)

    return np.clip(out, 0, 255).astype(np.uint8)
````

````python
#testing our function
blurred_rgb = average_blur_rgb(img_rgb, ksize=11)
plt.figure(figsize=(6,6))
plt.imshow(blurred_rgb)
plt.title('Average blur on rgb')
plt.axis('off')
plt.show()
````

````python
#Gaussian blur
#Implementing a Gaussian kernel generator
def gaussian_kernel(ksize, sigma):
   #generate a normalized gaussian kernel
   #gives highest value at center and rest value decreases with distance from center
    radius = ksize // 2
    x = np.arange(-radius, radius + 1) #[-r,-r+1...0,1,2...r]
    xx, yy = np.meshgrid(x, x) #creates a 2d grid for coordinates

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2)) #formula for gaussian blur
    kernel = kernel / kernel.sum() #normlaizing so that the sum is 1-brighntness remains same

    return kernel
````

````python
#convolving it with a grayscale image
def gaussian_blur_gray(image, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(image, kernel)
````

````python
gaussian_blur_gimg=gaussian_blur_gray(gray, ksize=11, sigma=4) #small sigma-less blur large sigma-more blur
plt.figure(figsize=(6,6))
plt.imshow(gaussian_blur_gimg, cmap='gray')
plt.title('Gaussian blur on greyscale')
plt.axis('off')
plt.show()
````

````python
#gaussian blur for rgb
def gaussian_blur_rgb(image, ksize, sigma):
    blurred = np.zeros_like(image, dtype=np.float64)

    for channel in range(3):
        blurred[..., channel] = gaussian_blur_gray(
            image[..., channel], ksize, sigma
        )

    return np.clip(blurred, 0, 255).astype(np.uint8)
````

````python
gaussian_blur_rgb_img=gaussian_blur_rgb(img_rgb, ksize=11, sigma=4)
plt.figure(figsize=(6,6))
plt.imshow(gaussian_blur_rgb_img)
plt.title('Gaussian blur on rgb')
plt.axis('off')
plt.show()
````

````python
#create a function for magnitude spectrum first
def magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude = np.log(1 + np.abs(fshift))
    return magnitude
````

````python
#getting the required images
# Original
orig = gray

# Gaussian blurred
blurred = gaussian_blur_gray(gray, ksize=5, sigma=1.0)

# Unsharp masking
alpha = 1.5
mask = gray - blurred
unsharp = gray + alpha * mask
unsharp = np.clip(unsharp, 0, 255)
````

````python
#computing magnitude spectrum
mag_orig = magnitude_spectrum(orig)
mag_blur = magnitude_spectrum(blurred)
mag_unsharp = magnitude_spectrum(unsharp)
````
