````python
#writing a function gray_histogram
def gray_histogram(gray):
  hist=np.zeros(256,dtype=np.int64) #hist=[0,0,0,0,.......0] of size 256
  h, w = gray.shape
  #running a for loop to calculate the number of pixels associated with each brightness value
  for i in range(h):
        for j in range(w):
            val = int(gray[i, j])
            hist[val] += 1

  return hist
````

````python
#plotting grayscale histogram
hist_gray = gray_histogram(gray)

plt.figure(figsize=(6,6))
plt.plot(np.arange(256), hist_gray)
plt.title("Grayscale Histogram")
plt.xlabel("Intensity")
plt.ylabel("Pixel Count")
plt.xlim(0, 255)
plt.show()
````

````python
#writing funtion rgb_histogram
def rgb_histogram(rgb_img):
    hist_r = np.zeros(256, dtype=np.int64)
    hist_g = np.zeros(256, dtype=np.int64)
    hist_b = np.zeros(256, dtype=np.int64)

    h, w, _ = rgb_img.shape

    for i in range(h):
        for j in range(w):
            r, g, b = rgb_img[i, j]
            hist_r[int(r)] += 1
            hist_g[int(g)] += 1
            hist_b[int(b)] += 1

    return hist_r, hist_g, hist_b
````

````python
#plotting rgb histogram
hist_r, hist_g, hist_b = rgb_histogram(img_rgb)

bins = np.arange(256)

plt.figure(figsize=(6,6))
plt.plot(bins, hist_r, label="Red")
plt.plot(bins, hist_g, label="Green")
plt.plot(bins, hist_b, label="Blue")
plt.title("RGB Histogram")
plt.xlabel("Intensity")
plt.ylabel("Pixel Count")
plt.xlim(0, 255)
plt.legend()
plt.show()
````

Write a function <code>rgb_to_hsv</code> that takes a 0-255 ( not normalised ) RGB image and returns an HSV image ( also not normalised ).

````python
import numpy as np

def rgb_to_hsv(rgb):
    # convert from integers (0–255) to floats (0–1)
    rgb_f = rgb.astype(np.float32) / 255.0
    #splitting rgb channels
    r, g, b = rgb_f[..., 0], rgb_f[..., 1], rgb_f[..., 2]

    #max and min between rgb for each pixel (helpful for saturation and hue)
    maxc = np.max(rgb_f, axis=-1)
    minc = np.min(rgb_f, axis=-1)
    delta = maxc - minc

    # Value
    v = maxc

    # Saturation
    #if pixel is black s=0
    #else s=delta/maxc
    s = np.zeros_like(maxc) #np.zeros_like basically gives a blank array the same size as given function
    nonzero_v = maxc != 0
    s[nonzero_v] = delta[nonzero_v] / maxc[nonzero_v]

    # Hue
    h = np.zeros_like(maxc)

    # Avoid division by zero
    nz = delta != 0

    # Masks to find the dominant channel
    mask_r = (maxc == r) & nz
    mask_g = (maxc == g) & nz
    mask_b = (maxc == b) & nz

    # hue formulas for each case
    h[mask_r] = ((g - b)[mask_r] / delta[mask_r]) % 6
    h[mask_g] = ((b - r)[mask_g] / delta[mask_g]) + 2
    h[mask_b] = ((r - g)[mask_b] / delta[mask_b]) + 4

    # Dividing by 6 to normalize hue into [0,1).
    h = h / 6.0

    # scaling to open cv ranges
    H = np.round(h * 179).astype(np.uint8)     # [0,179]
    S = np.round(s * 255).astype(np.uint8)     # [0,255]
    V = np.round(v * 255).astype(np.uint8)     # [0,255]

    hsv = np.stack([H, S, V], axis=-1)
    return hsv
````

````python
#testing with cv2.cvtColor
#converting rgb to hsv using my function
hsv_my = rgb_to_hsv(img_rgb)
#converting back to rgb using open cv
rgb_back = cv2.cvtColor(hsv_my, cv2.COLOR_HSV2RGB)
#plotting original image vs reconstructed one
plt.figure(figsize=(6,6))
plt.title("Original RGB")
plt.imshow(img_rgb)
plt.axis("off")
plt.show

plt.figure(figsize=(6,6))
plt.title("Reconstructed RGB")
plt.imshow(rgb_back)
plt.axis("off")
plt.show()
````

#FILTER 1-"DREAMY POP"
###Idea is to make images look extra vivid. A bit brighter, more contrasty and with boosted color intensity
###What the filter is doing:
### 1. Boosting brightness+contrast
### 2. Increasing saturation
### 3. Increasing vibrance

````python
import cv2
import numpy as np

def dreamy_pop(img_rgb):

    # Brightness + contrast
    alpha = 1.2   # contrast factor
    beta  = 20  # brightness factor
    img_cb = cv2.convertScaleAbs(img_rgb, alpha=alpha, beta=beta)

    # increasing saturation
    hsv = cv2.cvtColor(img_cb, cv2.COLOR_RGB2HSV).astype(np.float32)

    sat_scale = 1.2
    hsv[..., 1] = np.clip(hsv[..., 1] * sat_scale, 0, 255)

    # Extra vibrance
    s = hsv[..., 1] / 255.0
    boost = 0.7
    delta = (1.0 - s) * boost
    new_s = np.clip((s + delta * s), 0.0, 1.0)
    hsv[..., 1] = new_s * 255.0

    # Returning back to rgb
    out = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB)
    return out
````

````python
img_dreamy_pop=dreamy_pop(img_rgb)
plt.figure(figsize=(6,6))
plt.title("Dreamy Pop")
plt.imshow(img_dreamy_pop)
plt.axis("off")
plt.show()
````

````python
# I used brightness/contrast and saturation boost to make the colorful eye look even more vibrant.
# These changes help the colors stand out more and make the details look sharper.
# This filter works well on bright, colorful images because it enhances what’s already there.
````


#FILTER 2-"BLUETRON VISION"
###Idea is to make image look more dramatic/cinematic with a blue tint to it.
###What the filter is doing:
### 1. Reducing saturation
### 2. Increase contrast
### 3. Add a blue tint

````python
def bluetron_vision(img_rgb):
    # reducing saturation
    hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV).astype(np.float32)

    sat_scale = 0.3
    hsv[..., 1] = np.clip(hsv[..., 1] * sat_scale, 0, 255)

    # returning back to rgb
    low_sat = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB)

    # stronger contrast
    alpha = 1.4
    beta  = -10   #a bit of darkening
    high_contrast = cv2.convertScaleAbs(low_sat, alpha=alpha, beta=beta)

    # Adding blue tint
    #channel scaling
    tinted = high_contrast.astype(np.float32)

    tinted[..., 0] *= 0.6  # reducing red
    tinted[..., 1] *= 0.8   #reducing green
    tinted[..., 2] *= 1.4 # boosting blue
    #values can be changed according to the image..these are the values which i think worked best for mine
    tinted = np.clip(tinted, 0, 255).astype(np.uint8)

    return tinted
````

````python
img_bluetron_vision=bluetron_vision(img_rgb)
plt.figure(figsize=(6,6))
plt.title("Bluetron Vision")
plt.imshow(img_bluetron_vision)
plt.axis("off")
plt.show()
````

````python
# I lowered the saturation, increased the contrast, and added a blue tint to give the image a cool, dramatic look.
# This works well for the eye painting because the contrast makes the details stand out and the blue tone adds a mysterious feel.
````

#### White Balance Function
Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image. Use Numpy only to keep track of the brightest pixel and then scale the entire image with the factor you find. Display the original and output image.

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.
````

````python
import numpy as np

def white_patch_balance(img_rgb):

    img = img_rgb.astype(np.float32)

    # Find maximum value in each channel
    max_r = np.max(img[..., 0])
    max_g = np.max(img[..., 1])
    max_b = np.max(img[..., 2])

    #Compute scaling factors to change brightest pixel to 255
    scale_r = 255.0 / max_r
    scale_g = 255.0 / max_g
    scale_b = 255.0 / max_b

    # Apply scaling to each channel
    out = img.copy()
    out[..., 0] *= scale_r
    out[..., 1] *= scale_g
    out[..., 2] *= scale_b

    out = np.clip(out, 0, 255).astype(np.uint8)

    return out
````

````python
plt.figure(figsize=(6,6))
plt.title("Original Image")
plt.imshow(img_rgb)
plt.axis("off")
plt.show()

white_balance_img=white_patch_balance(img_rgb)
plt.figure(figsize=(6,6))
plt.title("White Balance Image")
plt.imshow(white_balance_img)
plt.axis("off")
plt.show()
````

It takes the img_rgb ( 0-255 RGB image ) and value ( 0-1 decimal number ).
1. value = 0 outputs the coldest version of the image

3. value = 1 is the warmest version of the image

````python
import numpy as np

def white_balance(img_rgb, value):
    img = img_rgb.astype(np.float32)

    # value = 0 : shift = -1 (cool)
    # value = 0.5 : shift = 0  (no change)
    # value = 1 : shift = +1 (warm)
    shift = (value - 0.5) * 2.0

    # Strength of color adjustment
    strength = 40

    #applying opposite adjustments to red and blue
    #on warm setting-red is boosted, blue is reduced
    #on cool setting-red is reduced, blue is boosted
    img[..., 0] += shift * strength      # Red channel
    img[..., 2] -= shift * strength      # Blue channel

    img = np.clip(img, 0, 255).astype(np.uint8)

    return img
````

````python
#trying the white balance slider
img_value_0=white_balance(img_rgb, 0)
img_value_half=white_balance(img_rgb, 0.5)
img_value_1=white_balance(img_rgb, 1)

plt.figure(figsize=(6,6))
plt.imshow(img_value_0)
plt.title("Value=0")
plt.axis('off')
plt.show()

plt.figure(figsize=(6,6))
plt.imshow(img_value_half)
plt.title("Value=0.5")
plt.axis('off')
plt.show()

plt.figure(figsize=(6,6))
plt.imshow(img_value_1)
plt.title("Value=1")
plt.axis('off')
plt.show()
````
