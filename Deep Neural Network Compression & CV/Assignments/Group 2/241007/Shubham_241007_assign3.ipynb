````python
def convolve(img, kernel):
    kh, kw = kernel.shape
    pad_h = kh // 2
    pad_w = kw // 2

    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    out = np.zeros_like(img, dtype=np.float32)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            region = padded[i:i+kh, j:j+kw]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float32)
    kernel /= (ksize * ksize)
    return convolve(img, kernel)
````

````python
avg_blur_gray = average_blur_gray(gray, 7)
plt.imshow(avg_blur_gray, cmap='gray')
plt.axis("off")
plt.show()
````

````python
def average_blur_rgb(img, ksize):
    out = np.zeros_like(img, dtype=np.float32)
    for c in range(3):
        out[..., c] = average_blur_gray(img[..., c], ksize)
    return np.clip(out, 0, 255).astype(np.uint8)
````

````python
avg_blur_rgb = average_blur_rgb(img_rgb, 7)
plt.imshow(avg_blur_rgb)
plt.axis("off")
plt.show()
````

````python
def gaussian_kernel(ksize, sigma):
    ax = np.arange(-ksize//2 + 1, ksize//2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= np.sum(kernel)
    return kernel
````

````python
def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)
````

````python
def gaussian_blur_rgb(img, ksize, sigma):
    out = np.zeros_like(img, dtype=np.float32)
    for c in range(3):
        out[..., c] = gaussian_blur_gray(img[..., c], ksize, sigma)
    return np.clip(out, 0, 255).astype(np.uint8)
````

````python
gauss_gray = gaussian_blur_gray(gray, 7, 1.5)
gauss_rgb = gaussian_blur_rgb(img_rgb, 7, 1.5)
````

Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.
````

````python
gauss = gaussian_kernel(9, 2.0)
blurred = convolve(gray, gauss)

mask = gray - blurred
alpha = 1.0
unsharp = gray + alpha * mask
````

````python
def magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    mag = np.log(1 + np.abs(fshift))
    return mag
````

````python
spec_original = magnitude_spectrum(gray)
spec_blur = magnitude_spectrum(blurred)
spec_unsharp = magnitude_spectrum(unsharp)
````

````python
plt.figure(figsize=(12,6))

plt.subplot(2,3,1); plt.title("Original Image"); plt.imshow(gray, cmap='gray'); plt.axis("off")
plt.subplot(2,3,2); plt.title("Gaussian Blurred"); plt.imshow(blurred, cmap='gray'); plt.axis("off")
plt.subplot(2,3,3); plt.title("Unsharp Masked"); plt.imshow(unsharp, cmap='gray'); plt.axis("off")

plt.subplot(2,3,4); plt.title("Original Spectrum"); plt.imshow(spec_original, cmap='gray'); plt.axis("off")
plt.subplot(2,3,5); plt.title("Blur Spectrum (LPF)"); plt.imshow(spec_blur, cmap='gray'); plt.axis("off")
plt.subplot(2,3,6); plt.title("Unsharp Spectrum (HPF)"); plt.imshow(spec_unsharp, cmap='gray'); plt.axis("off")

plt.tight_layout()
plt.show()
````

````python
# Observations:
# 1. In the original image spectrum, energy is concentrated near the center,
#    which corresponds to low-frequency components (smooth regions).
#
# 2. After Gaussian blur, the center becomes brighter and outer regions darker,
#    showing that high frequencies (edges, fine details) are suppressed.
#    This verifies Gaussian blur acts as a Low Pass Filter (LPF).
#
# 3. In the unsharp masking spectrum, the outer regions become brighter,
#    indicating amplification of high-frequency components.
#    This confirms unsharp masking enhances edges (High Pass behavior).
#
# 4. Increasing Gaussian kernel size or sigma further darkens outer frequencies.
#    Increasing alpha in unsharp masking increases high-frequency brightness.
````
