````python
#Task 1
import numpy as np

def convolve(img, kernel):
    # Flip kernel for convolution
    kernel = np.flip(kernel, axis=(0, 1))

    img_rows, img_cols = img.shape
    k_rows, k_cols = kernel.shape

    pad_r = k_rows // 2
    pad_c = k_cols // 2

    padded_img = np.pad(
        img,
        ((pad_r, pad_r), (pad_c, pad_c)),
        mode='reflect'
    )
    result = np.zeros((img_rows, img_cols), dtype=np.float64)

    for r in range(img_rows):
        for c in range(img_cols):
            region = padded_img[r:r + k_rows, c:c + k_cols]
            result[r, c] = np.sum(region * kernel)
    return result
````

````python
import numpy as np
import matplotlib.pyplot as plt

def average_blur_gray(img, ksize):
    assert ksize % 2 == 1, "ksize must be odd"
    rows, cols = img.shape
    pad = ksize // 2

    box_kernel = np.full(
        (ksize, ksize),
        1.0 / (ksize * ksize),
        dtype=np.float64
    )

    padded = np.pad(
        img,
        ((pad, pad), (pad, pad)),
        mode="reflect"
    )

    output = np.zeros((rows, cols), dtype=np.float64)

    for i in range(rows):
        for j in range(cols):
            region = padded[i:i + ksize, j:j + ksize]
            output[i, j] = np.sum(region * box_kernel)

    return output

gray_path = "/content/drive/MyDrive/dog.jpg"
gray_img = plt.imread(gray_path).astype(np.float64)

if gray_img.max() > 1.0:
    gray_img /= 255.0

if gray_img.ndim == 3:
    gray_img = (
        0.299 * gray_img[:, :, 0] +
        0.587 * gray_img[:, :, 1] +
        0.114 * gray_img[:, :, 2]
    )

gray_blurred = average_blur_gray(gray_img, ksize=5)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_blurred, cmap="gray")
plt.title("Average Blurred Image")
plt.axis("off")

plt.show()

def average_blur_rgb(img, ksize):
    assert ksize % 2 == 1, "ksize must be odd"

    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:, :, :3]

    img = img.astype(np.float64)

    if img.max() > 1.0:
        img /= 255.0

    h, w, _ = img.shape
    blurred_rgb = np.zeros_like(img, dtype=np.float64)

    for c in range(3):
        blurred_rgb[:, :, c] = average_blur_gray(img[:, :, c], ksize)

    return blurred_rgb


rgb_path = "/content/drive/MyDrive/heli.jpg"
rgb_img = plt.imread(rgb_path)

if rgb_img.ndim == 3 and rgb_img.shape[2] == 4:
    rgb_img = rgb_img[:, :, :3]

rgb_blurred = average_blur_rgb(rgb_img, ksize=7)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(rgb_img)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(rgb_blurred)
plt.title("Average Blurred RGB")
plt.axis("off")

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

def gaussian_kernel(ksize, sigma):
    assert ksize % 2 == 1, "ksize must be odd"

    offset = ksize // 2
    axis = np.arange(-offset, offset + 1)
    xx, yy = np.meshgrid(axis, axis)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= kernel.sum()
    return kernel


def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)


def gaussian_blur_rgb(img, ksize, sigma):
    assert ksize % 2 == 1, "ksize must be odd"

    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:, :, :3]

    img = img.astype(np.float64)

    if img.max() > 1.0:
        img /= 255.0

    hgt, wdt, _ = img.shape
    out = np.zeros((hgt, wdt, 3), dtype=np.float64)
    kernel = gaussian_kernel(ksize, sigma)

    for c in range(3):
        out[:, :, c] = convolve(img[:, :, c], kernel)

    return out


img_file = "/content/drive/MyDrive/heli.jpg"
raw_img = plt.imread(img_file).astype(np.float64)

if raw_img.max() > 1.0:
    raw_img /= 255.0

rgb_img = raw_img.copy()

if raw_img.ndim == 3:
    gray_img = (
        0.299 * raw_img[:, :, 0] +
        0.587 * raw_img[:, :, 1] +
        0.114 * raw_img[:, :, 2]
    )
else:
    gray_img = raw_img


ksize = 7
sigma = 1.5


gray_blurred = gaussian_blur_gray(gray_img, ksize, sigma)
rgb_blurred = gaussian_blur_rgb(rgb_img, ksize, sigma)



plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray_img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_blurred, cmap="gray")
plt.title("Gaussian Blurred Grayscale")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(rgb_img)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(rgb_blurred)
plt.title("Gaussian Blurred RGB")
plt.axis("off")

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Sobel operators
sobel_x = np.array(
    [[-1,  0,  1],
     [-2,  0,  2],
     [-1,  0,  1]],
    dtype=np.float64
)

sobel_y = np.array(
    [[-1, -2, -1],
     [ 0,  0,  0],
     [ 1,  2,  1]],
    dtype=np.float64
)

# Load image
img_file = "/content/drive/MyDrive/heli.jpg"
img_data = plt.imread(img_file)

if img_data.ndim == 3:
    gray = img_data[:, :, :3].mean(axis=2)
else:
    gray = img_data
gray = gray.astype(np.float64)
# Apply Sobel filters
gx = convolve(gray, sobel_x)
gy = convolve(gray, sobel_y)
# Gradient magnitude
magnitude = np.sqrt(gx**2 + gy**2)
# Normalize to 0–255
norm_mag = (magnitude / magnitude.max()) * 255
norm_mag = norm_mag.astype(np.uint8)
# Thresholding
threshold = 50
edge_map = np.zeros_like(norm_mag)
edge_map[norm_mag > threshold] = 255
# Display
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gx, cmap="gray")
plt.title("Sobel X Response")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gy, cmap="gray")
plt.title("Sobel Y Response")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(norm_mag, cmap="gray")
plt.title("Gradient Magnitude (Normalized)")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(edge_map, cmap="gray")
plt.title("Binary Edge Map")
plt.axis("off")

plt.show()
````
