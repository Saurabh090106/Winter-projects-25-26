````python
#Task 1
import numpy as np

def convolve(img, kernel):
    # Reverse the kernel
    kernel = np.flipud(np.fliplr(kernel))

    img_h, img_w = img.shape
    ker_h, ker_w = kernel.shape
    # Amount of padding needed
    pad_y = ker_h // 2
    pad_x = ker_w // 2
    # Pad image using reflection
    padded_img = np.pad(
        img,
        ((pad_y, pad_y), (pad_x, pad_x)),
        mode='reflect'
    )
    # Store result
    conv_result = np.zeros((img_h, img_w), dtype=np.float64)
    # Slide kernel over image
    for y in range(img_h):
        for x in range(img_w):
            region = padded_img[y:y+ker_h, x:x+ker_w]
            conv_result[y, x] = np.sum(region * kernel)
    return conv_result
````

````python
#Task 2
import numpy as np
import matplotlib.pyplot as plt

def average_blur_gray(img, ksize):
    assert ksize % 2 == 1, "ksize must be odd"
    h, w = img.shape
    border = ksize // 2
    # Uniform averaging kernel
    avg_kernel = np.ones((ksize, ksize), dtype=np.float64) / (ksize * ksize)
    # Mirror padding
    padded_img = np.pad(
        img,
        ((border, border), (border, border)),
        mode="reflect"
    )
    # Output image
    blurred_img = np.zeros((h, w), dtype=np.float64)
    # Apply kernel over image
    for y in range(h):
        for x in range(w):
            patch = padded_img[y:y+ksize, x:x+ksize]
            blurred_img[y, x] = np.sum(patch * avg_kernel)
    return blurred_img

# Load grayscale image
img_path = "/content/drive/MyDrive/Screenshot 2025-12-15 190829.png"
gray_img = plt.imread(img_path)

if gray_img.ndim == 3:
    gray_img = gray_img.mean(axis=2)

gray_img = gray_img.astype(np.float64)
gray_blur = average_blur_gray(gray_img, ksize=5)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_blur, cmap="gray")
plt.title("Average Blurred Image")
plt.axis("off")
plt.show()

def average_blur_rgb(img, ksize):
    assert ksize % 2 == 1, "ksize must be odd"
    # Drop alpha channel if present
    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:, :, :3]

    height, width, channels = img.shape
    assert channels == 3, "Input must be RGB"
    img = img.astype(np.float64)
    rgb_out = np.zeros_like(img, dtype=np.float64)
    # Blur each color channel separately
    for ch in range(3):
        rgb_out[:, :, ch] = average_blur_gray(img[:, :, ch], ksize)
    return rgb_out

# Load RGB image
img_path = "/content/drive/MyDrive/Screenshot 2025-12-15 211230.png"
color_img = plt.imread(img_path)

if color_img.ndim == 3 and color_img.shape[2] == 4:
    color_img = color_img[:, :, :3]

rgb_blurred = average_blur_rgb(color_img, ksize=7)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(np.clip(color_img, 0, 1))
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(np.clip(rgb_blurred, 0, 1))
plt.title("Average Blurred RGB")
plt.axis("off")

plt.show()
````

````python
#Task 3
import numpy as np
import matplotlib.pyplot as plt

def gaussian_kernel(ksize, sigma):
    assert ksize % 2 == 1, "ksize must be odd"
    # Coordinate values centered at zero
    coords = np.arange(-(ksize // 2), ksize // 2 + 1)
    X, Y = np.meshgrid(coords, coords)
    # Gaussian function
    gauss = np.exp(-(X**2 + Y**2) / (2 * sigma**2))
    # Normalize kernel
    gauss /= np.sum(gauss)
    return gauss

def gaussian_blur_gray(img, ksize, sigma):
    g_kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, g_kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    assert ksize % 2 == 1, "ksize must be odd"
    # Remove alpha channel if present
    if img.ndim == 3 and img.shape[2] == 4:
        img = img[:, :, :3]

    img = img.astype(np.float64)
    h, w, ch = img.shape
    assert ch == 3, "Input must be RGB"
    blurred_img = np.zeros_like(img, dtype=np.float64)
    g_kernel = gaussian_kernel(ksize, sigma)
    # Process each color channel separately
    for idx in range(3):
        blurred_img[:, :, idx] = convolve(img[:, :, idx], g_kernel)
    return blurred_img

# Load and prepare image
img_path = "/content/drive/MyDrive/Screenshot 2025-12-15 211230.png"
loaded_img = plt.imread(img_path)

rgb_img = loaded_img.copy()
# Convert to grayscale if needed
if loaded_img.ndim == 3:
    gray_img = loaded_img[:, :, :3].mean(axis=2)
else:
    gray_img = loaded_img

gray_img = gray_img.astype(np.float64)
# Parameters
ksize = 7
sigma = 1.5
# Apply Gaussian blur
gray_result = gaussian_blur_gray(gray_img, ksize, sigma)
rgb_result  = gaussian_blur_rgb(rgb_img, ksize, sigma)
# Display results
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray_img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_result, cmap="gray")
plt.title("Gaussian Blurred Grayscale")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(np.clip(rgb_img, 0, 1))
plt.title("Original RGB")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(np.clip(rgb_result, 0, 1))
plt.title("Gaussian Blurred RGB")
plt.axis("off")

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Sobel kernels for horizontal and vertical edges
sobel_h = np.array([
    [-1,  0,  1],
    [-2,  0,  2],
    [-1,  0,  1]
], dtype=np.float64)

sobel_v = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float64)

# Load image
img_path = "/content/drive/MyDrive/Screenshot 2025-12-15 211230.png"
raw_img = plt.imread(img_path)
# Convert to grayscale
if raw_img.ndim == 3:
    gray_img = raw_img[:, :, :3].mean(axis=2)
else:
    gray_img = raw_img
gray_img = gray_img.astype(np.float64)
# Apply Sobel filters
grad_x = convolve(gray_img, sobel_h)
grad_y = convolve(gray_img, sobel_v)
# Gradient magnitude
grad_mag = np.sqrt(grad_x**2 + grad_y**2)
# Scale range to 0–255
grad_scaled = (grad_mag / grad_mag.max()) * 255
grad_scaled = grad_scaled.astype(np.uint8)
# Simple thresholding
thresh_val = 50
edges = np.zeros_like(grad_scaled)
edges[grad_scaled > thresh_val] = 255
# Display results
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(gray_img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(grad_x, cmap="gray")
plt.title("Sobel X Response")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(grad_y, cmap="gray")
plt.title("Sobel Y Response")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(grad_scaled, cmap="gray")
plt.title("Normalized Gradient Magnitude")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(edges, cmap="gray")
plt.title("Binary Edge Map")
plt.axis("off")

plt.show()
````

````python
#Task 1
import cv2
import matplotlib.pyplot as plt

# Laplacian mask (4-connected)
laplacian_kernel = np.array([
    [ 0, -1,  0],
    [-1,  4, -1],
    [ 0, -1,  0]
], dtype=np.float64)

# Load image
img_path = "/content/drive/MyDrive/Screenshot 2025-12-15 211230.png"
input_img = plt.imread(img_path)
# Convert to grayscale if RGB
if input_img.ndim == 3:
    gray = input_img[:, :, :3].mean(axis=2)
else:
    gray = input_img
gray = gray.astype(np.float64)
# Laplacian using custom convolution
lap = convolve(gray, laplacian_kernel)
# Sharpening with different alpha values
alpha = 0.5
sharp = gray - lap * alpha

alpha = 1.0
sharp_numpy_alpha1 = gray - lap * alpha
# OpenCV Laplacian for comparison
lap_cv = cv2.Laplacian(gray, cv2.CV_64F, ksize=3)
sharp_cv = gray - lap_cv
# Difference between numpy and OpenCV result
diff_img = np.abs(sharp_numpy_alpha1 - sharp_cv)
max_diff = np.max(diff_img)
print("Maximum absolute difference:", max_diff)
# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(lap, cmap="gray")
plt.title("Laplacian Response")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(sharp_numpy_alpha1, cmap="gray")
plt.title("Sharpened (NumPy, α=1)")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(sharp_cv, cmap="gray")
plt.title("Sharpened (OpenCV)")
plt.axis("off")

plt.show()
````

````python
# Gaussian filter parameters
kernel_size = 9
std_dev = 2.0
# Build Gaussian kernel
g_kernel = gaussian_kernel(kernel_size, std_dev)
# Low-pass filtering using Gaussian blur
smooth_img = convolve(gray, g_kernel)

alpha = 1.5

# High-frequency component
mask = gray - smooth_img
# Unsharp masking result
sharp_img = gray + alpha * mask

def magnitude_spectrum(image):
    # Compute 2D FFT
    fft_img = np.fft.fft2(image)
    # Shift zero frequency component to center
    fft_centered = np.fft.fftshift(fft_img)
    # Log-scaled magnitude
    spectrum = np.log(1 + np.abs(fft_centered))
    return spectrum
# Frequency domain representations
spec_orig  = magnitude_spectrum(gray)
spec_smooth = magnitude_spectrum(smooth_img)
spec_sharp = magnitude_spectrum(sharp_img)
# Display spatial and frequency results
plt.figure(figsize=(15, 10))

plt.subplot(3, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(3, 2, 2)
plt.imshow(spec_orig, cmap="gray")
plt.title("Spectrum of Original")
plt.axis("off")

plt.subplot(3, 2, 3)
plt.imshow(smooth_img, cmap="gray")
plt.title("Gaussian Smoothed (Low-Pass)")
plt.axis("off")

plt.subplot(3, 2, 4)
plt.imshow(spec_smooth, cmap="gray")
plt.title("Spectrum of Smoothed Image")
plt.axis("off")

plt.subplot(3, 2, 5)
plt.imshow(sharp_img, cmap="gray")
plt.title("Unsharp Masking Output")
plt.axis("off")

plt.subplot(3, 2, 6)
plt.imshow(spec_sharp, cmap="gray")
plt.title("Spectrum of Sharpened Image")
plt.axis("off")

plt.show()
````
