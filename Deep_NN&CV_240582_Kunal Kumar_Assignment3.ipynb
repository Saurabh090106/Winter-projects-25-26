````python
#image loading
import cv2
import numpy as np
import matplotlib.pyplot as plt

org_img=cv2.imread("/content/1st_image_for_assignment_3.jpg")
gray_img=cv2.cvtColor(org_img,cv2.COLOR_BGR2GRAY)
rgb_img=cv2.cvtColor(org_img,cv2.COLOR_BGR2RGB)
plt.figure(figsize=(15,10))
plt.subplot(1,2,1)
plt.imshow(gray_img,cmap='gray')
plt.title("Original image in Grayscale")

plt.subplot(1,2,2)
plt.imshow(rgb_img)
plt.title("RGB Image")
plt.show()
````

````python
#A
#Task 1
def convolve(img,kernel):

  if len(img.shape)==3:
    #this is for the rgb channels
    height,width,channels=img.shape
    res=np.zeros_like(img, dtype=np.float64)
    for j in range(channels):
      res[:,:,j]=convolve(img[:,:,j],kernel)
    return res

  img_height,img_width=img.shape
  kernel_height,kernel_width=kernel.shape
  pad_h=kernel_height//2
  pad_w=kernel_width//2
  padded_image=np.pad(img,((pad_h,pad_h),(pad_w,pad_w)),mode='reflect')

  res=np.zeros_like(img,dtype=np.float64)
  for i in range(img_height):
    for j in range(img_width):
      r=padded_image[i:i+kernel_height,j:j+kernel_width]
      res[i,j]=np.sum(r*kernel)

  return res

# kernel=np.array([[1,2,1],[2,4,2],[1,2,1]])/16
# result=convolve(gray_img,kernel)

# plt.imshow(result,cmap='gray')
# plt.title("Convolution Output")
# plt.show()
````

````python
#Task 2
def average_blur(img,ksize):
  kernel=np.ones((ksize,ksize))/(ksize*ksize)
  res=convolve(img,kernel)
  return np.clip(res,0,255).astype(np.uint8)
avg_gray=average_blur(gray_img,ksize=9)
avg_rgb=average_blur(rgb_img,ksize=9)

plt.figure(figsize=(12,10))
#Grayscale plot
plt.subplot(2,2,1)
plt.imshow(gray_img,cmap='gray')
plt.title(f"Original Grayscale Image")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(avg_gray, cmap='gray')
plt.title(f"Average Blur (Gray)\nksize=9")
plt.axis('off')

# RGB Plot
plt.subplot(2,2,3)
plt.imshow(rgb_img)
plt.title(f"Original RGB_Image")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(avg_rgb)
plt.title(f"Average Blur (RGB)\nksize=9")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
#Task 3
def gaussian_kernel(ksize,sigma):
  kernel=np.zeros((ksize,ksize))
  centre=ksize//2
  for i in range(ksize):
    for j in range(ksize):
      x=i-centre; y=j-centre;
      v=-(x*x+y*y)/(2*sigma*sigma)
      kernel[i,j]=np.exp(v)
  return kernel/np.sum(kernel)

# now applying the gaussian kernel
def apply(img,ksize,sigma):
  kernel=gaussian_kernel(ksize,sigma)
  res=convolve(img,kernel)
  return np.clip(res,0,255).astype(np.uint8)
#1 on gray image
# the sigma i use here in order to get [[1 2 1],[2 4 2],[1 2 1]]/16 this is 0.85
gauss_gray=apply(gray_img,3,0.85)

#2nd on rgb image
# i have used ksize=9 and sigma 2 to enhance the blur in rgb
# this i used ksize=5 and sigma 2.0 in order to increase the blur effect
gauss_rgb=apply(rgb_img,5,2.0)

plt.figure(figsize=(12,10))
#Gray
plt.subplot(2,2,1)
plt.imshow(gray_img,cmap='gray')
plt.title(f"Original Gray Image")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(gauss_gray,cmap='gray')
plt.title(r"Gaussian Blur (Gray), ksize=3, $\sigma$=0.85")
plt.axis('off')

#RGB
plt.subplot(2,2,3)
plt.imshow(rgb_img)
plt.title(f"Original RGB Image")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(gauss_rgb)
plt.title(r"Gaussian Blur (RGB), ksize=5, $\sigma$=2")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
def get_magnitude_spectrum(img):
    f=np.fft.fft2(img)
    fshift=np.fft.fftshift(f)
    magnitude=np.abs(fshift)
    return 20*np.log(magnitude+1)

#Blur
ksize=9 # further increased the ksize in order to increase the blur effect again
sigma=2
blurr_img=apply(gray_img,ksize,sigma)
#sharpening
alpha=2.0
mask=gray_img.astype(float)-blurr_img.astype(float)
sharpened_img=gray_img+(alpha*mask)
sharpened_img=np.clip(sharpened_img,0,255).astype(np.uint8)
#FFT
original_spectrum=get_magnitude_spectrum(gray_img)
blurred_spectrum=get_magnitude_spectrum(blurr_img)
sharpened_spectrum=get_magnitude_spectrum(sharpened_img)

#display
plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1)
plt.imshow(gray_img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(original_spectrum, cmap='gray')
plt.title("Original Spectrum")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(blurr_img, cmap='gray')
plt.title(f"Gaussian Blur (LPF)\nKernel:{ksize}x{ksize}")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(blurred_spectrum, cmap='gray')
plt.title("Blurred Spectrum (LPF Effect)")
plt.axis('off')


plt.subplot(2, 3, 3)
plt.imshow(sharpened_img, cmap='gray')
plt.title(f"Unsharp Masking (HPF)\nAlpha:{alpha}")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(sharpened_spectrum, cmap='gray')
plt.title("Sharpened Spectrum (HPF Boost)")
plt.axis('off')

plt.show()

#Observations and Frequency Analysis :
#Bright Areas: Represent Strong freqnecies. If a pixel in the spectrum is white, it means that specific pattern(frequency) is heavily present in the image
#Dark Areas: Represent weak and missing frrequencies. If an area is black,it means those patterns are essentially not in the image.
````

````python
def rgb_to_hsv(img_rgb):
    img = img_rgb.astype(np.float32) / 255.0
    r, g, b = img[..., 0], img[..., 1], img[..., 2]
    cmax=np.max(img, axis=2)
    cmin=np.min(img, axis=2)
    delta=cmax - cmin
    # calculation of hue
    h = np.zeros_like(cmax)
    mask = delta > 0
    # HSV Hue formulas
    idx=(cmax==r)&mask
    h[idx]=((g[idx]-b[idx])/delta[idx])%6
    idx=(cmax==g)&mask
    h[idx]=((b[idx]-r[idx])/delta[idx]) + 2
    idx=(cmax==b)&mask
    h[idx]=((r[idx]-g[idx])/delta[idx]) + 4
    #convertinf into degres
    h *= 60
    # OpenCV scaling
    h_cv=(h/2).astype(np.uint8)
    # Saturation Calculation
    s=np.zeros_like(cmax)
    idx=cmax>0
    s[idx]=delta[idx]/cmax[idx]
    #scaling
    s_cv=(s*255).astype(np.uint8)
    v_cv = (cmax * 255).astype(np.uint8)
    return np.dstack([h_cv, s_cv, v_cv])

my_hsv=rgb_to_hsv(img2)
h=my_hsv[:, :, 0]
s=my_hsv[:, :, 1]
v=my_hsv[:, :, 2]
mask_yellow=(h>15)&(h<30)&(s>50)
h[mask_yellow]=0
final_img=np.dstack([h,s,v])
final_img=cv2.cvtColor(final_img,cv2.COLOR_HSV2RGB)

plt.figure(figsize=(15,10))
plt.subplot(1,2,1)
plt.imshow(img2)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(final_img)
plt.title("New Image (Yellow Leaf -> Red Leaf)")
plt.axis('off')
plt.show()
````
