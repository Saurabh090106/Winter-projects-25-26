````python
import numpy as np
import cv2
import matplotlib.pyplot as plt


def convolve(image, kernel):
    h, w = image.shape
    k = kernel.shape[0] // 2
    padded = np.pad(image, k, mode='edge')
    out = np.zeros_like(image, dtype=np.float32)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+2*k+1, j:j+2*k+1]
            out[i, j] = np.sum(region * kernel)

    return out


laplacian_kernel = np.array([
    [ 0, -1,  0],
    [-1,  4, -1],
    [ 0, -1,  0]
], dtype=np.float32)


lap = convolve(gray, laplacian_kernel)


alpha = 0.5
sharp = gray - alpha * lap
sharp = np.clip(sharp, 0, 255).astype(np.uint8)


plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray')
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(lap, cmap='gray')
plt.title("Laplacian")

plt.subplot(1,3,3)
plt.imshow(sharp, cmap='gray')
plt.title(f"Sharpened (alpha={alpha})")
plt.axis("off")


alpha = 1.0
sharp_np = gray - lap
sharp_np = np.clip(sharp_np, 0, 255).astype(np.uint8)


lap_cv2 = cv2.Laplacian(gray, cv2.CV_32F)
sharp_cv2 = gray - lap_cv2
sharp_cv2 = np.clip(sharp_cv2, 0, 255).astype(np.uint8)

difference = np.abs(sharp_np.astype(int) - sharp_cv2.astype(int))
print("Max difference (NumPy vs cv2):", difference.max())
````

### Part D

#### Task - Frequency Analysis (LPF & HPF Verification)
Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    return np.log(1 + np.abs(fshift))

mag_original = magnitude_spectrum(gray)

gaussian_kernel = (15, 15)
blurred = cv2.GaussianBlur(gray, gaussian_kernel, 0)
mag_blur = magnitude_spectrum(blurred)

alpha = 1.5
mask = gray - blurred
unsharp = gray + alpha * mask
unsharp = np.clip(unsharp, 0, 255).astype(np.uint8)
mag_unsharp = magnitude_spectrum(unsharp)

plt.figure(figsize=(14,8))

plt.subplot(2,3,1)
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(blurred, cmap='gray')
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(unsharp, cmap='gray')
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(mag_original, cmap='gray')
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(mag_blur, cmap='gray')
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(mag_unsharp, cmap='gray')
plt.axis("off")

plt.tight_layout()
plt.show()
````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

R = img_np[:,:,0].astype(np.float32)
G = img_np[:,:,1].astype(np.float32)
B = img_np[:,:,2].astype(np.float32)

kernel = np.array([
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
], dtype=np.float32)

def convolve(image, kernel):
    h, w = image.shape
    k = kernel.shape[0] // 2
    padded = np.pad(image, k, mode='edge')
    out = np.zeros_like(image, dtype=np.float32)
    for i in range(h):
        for j in range(w):
            out[i,j] = np.sum(padded[i:i+2*k+1, j:j+2*k+1] * kernel)
    return out

R_edge = convolve(R, kernel)
G_edge = convolve(G, kernel)

RG_diff = R - G
RG_edge = convolve(RG_diff, kernel)

combined = 0.6 * R_edge + 0.6 * G_edge + 1.2 * RG_edge
combined = np.abs(combined)
combined = np.clip(combined, 0, 255).astype(np.uint8)

_, binary = cv2.threshold(combined, 40, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img_np)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(combined, cmap='gray')
plt.title("Colour Edge Response")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(binary, cmap='gray')
plt.title("Yellow Leaf Edge Outline")
plt.axis("off")

plt.show()
````
