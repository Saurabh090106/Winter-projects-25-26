````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

def rgb2gray(rgb_img):
    return np.dot(rgb_img[..., :3], [0.299, 0.587, 0.114])

def convolve(img, kernel):
    img_h, img_w = img.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(img)

    for i in range(img_h):
        for j in range(img_w):
            roi = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(roi * kernel)

    return output

def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize)) / (ksize * ksize)
    return convolve(img, kernel)

def average_blur_rgb(img, ksize):
    kernel = np.ones((ksize, ksize)) / (ksize * ksize)
    r_channel = img[:, :, 0]
    g_channel = img[:, :, 1]
    b_channel = img[:, :, 2]

    r_blurred = convolve(r_channel, kernel)
    g_blurred = convolve(g_channel, kernel)
    b_blurred = convolve(b_channel, kernel)

    return np.dstack((r_blurred, g_blurred, b_blurred))

def gaussian_kernel(ksize, sigma):
    pad = ksize // 2
    x_range = np.arange(-pad, pad + 1)
    y_range = np.arange(-pad, pad + 1)
    x_grid, y_grid = np.meshgrid(x_range, y_range)

    exponent = -(x_grid**2 + y_grid**2) / (2 * sigma**2)
    kernel_2d = np.exp(exponent)
    return kernel_2d / np.sum(kernel_2d)

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    r_channel = img[:, :, 0]
    g_channel = img[:, :, 1]
    b_channel = img[:, :, 2]

    r_blurred = convolve(r_channel, kernel)
    g_blurred = convolve(g_channel, kernel)
    b_blurred = convolve(b_channel, kernel)

    return np.dstack((r_blurred, g_blurred, b_blurred))

try:
    input_image = mpimg.imread('image.jpg')
    if input_image.dtype == np.uint8:
        input_image = input_image.astype(float) / 255.0
except FileNotFoundError:
    print("File 'image.jpg' not found. Please ensure the image is in the same directory.")
    exit()

gray_image = rgb2gray(input_image)

kernel_size = 15
sigma_val = 3

avg_blurred_gray = average_blur_gray(gray_image, kernel_size)
avg_blurred_rgb = average_blur_rgb(input_image, kernel_size)

gauss_kernel_matrix = gaussian_kernel(kernel_size, sigma_val)
gauss_blurred_gray = gaussian_blur_gray(gray_image, kernel_size, sigma_val)
gauss_blurred_rgb = gaussian_blur_rgb(input_image, kernel_size, sigma_val)

avg_blurred_gray = np.clip(avg_blurred_gray, 0, 1)
avg_blurred_rgb = np.clip(avg_blurred_rgb, 0, 1)
gauss_blurred_gray = np.clip(gauss_blurred_gray, 0, 1)
gauss_blurred_rgb = np.clip(gauss_blurred_rgb, 0, 1)

fig, axes = plt.subplots(3, 2, figsize=(12, 18))

axes[0, 0].imshow(input_image)
axes[0, 0].set_title("Original RGB")
axes[0, 0].axis('off')

axes[0, 1].imshow(gray_image, cmap='gray')
axes[0, 1].set_title("Original Gray")
axes[0, 1].axis('off')

axes[1, 0].imshow(avg_blurred_rgb)
axes[1, 0].set_title(f"Average Blur RGB (k={kernel_size})")
axes[1, 0].axis('off')

axes[1, 1].imshow(avg_blurred_gray, cmap='gray')
axes[1, 1].set_title(f"Average Blur Gray (k={kernel_size})")
axes[1, 1].axis('off')

axes[2, 0].imshow(gauss_blurred_rgb)
axes[2, 0].set_title(f"Gaussian Blur RGB (k={kernel_size}, sigma={sigma_val})")
axes[2, 0].axis('off')

axes[2, 1].imshow(gauss_blurred_gray, cmap='gray')
axes[2, 1].set_title(f"Gaussian Blur Gray (k={kernel_size}, sigma={sigma_val})")
axes[2, 1].axis('off')

plt.tight_layout()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

def rgb2gray(rgb_img):
    return np.dot(rgb_img[..., :3], [0.299, 0.587, 0.114])

def convolve(img, kernel):
    img_h, img_w = img.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(img)

    for i in range(img_h):
        for j in range(img_w):
            roi = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(roi * kernel)

    return output

try:
    input_image = mpimg.imread('image.jpg')
except FileNotFoundError:
    print("File 'image.jpg' not found.")
    exit()

if len(input_image.shape) == 3:
    gray_image = rgb2gray(input_image)
else:
    gray_image = input_image

sobel_x_kernel = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
])

sobel_y_kernel = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
])

gx = convolve(gray_image, sobel_x_kernel)
gy = convolve(gray_image, sobel_y_kernel)

gradient_magnitude = np.sqrt(gx**2 + gy**2)

if gradient_magnitude.max() > 0:
    normalized_magnitude = (gradient_magnitude / gradient_magnitude.max()) * 255
else:
    normalized_magnitude = gradient_magnitude
    
edge_map = normalized_magnitude.astype(np.uint8)

# I am choosing a threshold of 50 to filter out weak edges and keep strong ones.
# Values below 50 become 0 (black), and values above remain (or become 255 for binary).
threshold_value = 50
thresholded_map = np.where(edge_map > threshold_value, 255, 0)

plt.imsave('edge_map.png', thresholded_map, cmap='gray')

fig, axes = plt.subplots(2, 2, figsize=(10, 10))

axes[0, 0].imshow(gx, cmap='gray')
axes[0, 0].set_title("Sobel X (Gx)")
axes[0, 0].axis('off')

axes[0, 1].imshow(gy, cmap='gray')
axes[0, 1].set_title("Sobel Y (Gy)")
axes[0, 1].axis('off')

axes[1, 0].imshow(edge_map, cmap='gray')
axes[1, 0].set_title("Gradient Magnitude (Normalized)")
axes[1, 0].axis('off')

axes[1, 1].imshow(thresholded_map, cmap='gray')
axes[1, 1].set_title(f"Thresholded Edge Map (> {threshold_value})")
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
````

#### Task - Frequency Analysis (LPF & HPF Verification)
Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

def rgb2gray(rgb_img):
    return np.dot(rgb_img[..., :3], [0.299, 0.587, 0.114])

def convolve(img, kernel):
    img_h, img_w = img.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(img)

    for i in range(img_h):
        for j in range(img_w):
            roi = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(roi * kernel)

    return output

def gaussian_kernel(ksize, sigma):
    pad = ksize // 2
    x = np.arange(-pad, pad + 1)
    y = np.arange(-pad, pad + 1)
    x_grid, y_grid = np.meshgrid(x, y)
    exponent = -(x_grid**2 + y_grid**2) / (2 * sigma**2)
    kernel = np.exp(exponent)
    return kernel / np.sum(kernel)

def get_magnitude_spectrum(img):
    f = np.fft.fft2(img)
    
    fshift = np.fft.fftshift(f)
   
    # We add 1 to avoid log(0) errors
    magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)
    return magnitude_spectrum

# logic

try:
    input_image = mpimg.imread('image2.png')
except FileNotFoundError:
    print("File 'image2.png' not found.")
    exit()

if len(input_image.shape) == 3:
    gray_image = rgb2gray(input_image)
else:
    gray_image = input_image

# 1. Gaussian Blur (Low Pass Filter)
ksize = 9
sigma = 3.0
gauss_k = gaussian_kernel(ksize, sigma)
blurred_image = convolve(gray_image, gauss_k)

# 2. Unsharp Masking (Sharpening / High Freq Boost)
mask = gray_image - blurred_image
alpha = 2.0  
sharpened_image = gray_image + (alpha * mask)
sharpened_image = np.clip(sharpened_image, 0, 1)

# 3. Compute Spectrums
original_spectrum = get_magnitude_spectrum(gray_image)
blurred_spectrum = get_magnitude_spectrum(blurred_image)
sharpened_spectrum = get_magnitude_spectrum(sharpened_image)

# displaying results 

fig, axes = plt.subplots(3, 2, figsize=(12, 18))

axes[0, 0].imshow(gray_image, cmap='gray')
axes[0, 0].set_title("Original Image")
axes[0, 0].axis('off')

axes[0, 1].imshow(original_spectrum, cmap='gray')
axes[0, 1].set_title("Original Magnitude Spectrum")
axes[0, 1].axis('off')

axes[1, 0].imshow(blurred_image, cmap='gray')
axes[1, 0].set_title(f"Gaussian Blurred (k={ksize}, sigma={sigma})")
axes[1, 0].axis('off')

axes[1, 1].imshow(blurred_spectrum, cmap='gray')
axes[1, 1].set_title("Blurred Spectrum (Low Pass Filter)")
axes[1, 1].axis('off')

# OBSERVATION COMMENT FOR BLUR:
# In the spectrum above (Row 2, Right), look at the center vs the corners.
# The CENTER represents Low Frequencies (general shapes, smooth areas).
# The OUTER CORNERS represent High Frequencies (edges, noise, details).
# NOTICE: The outer areas are DARKER compared to the Original Spectrum.
# EXPLANATION: The Gaussian kernel is a Low Pass Filter. It passes low frequencies
# but blocks/attenuates high frequencies. Since edges/details were removed
# in the blurred image, their corresponding high frequencies in the spectrum are gone (dark).

axes[2, 0].imshow(sharpened_image, cmap='gray')
axes[2, 0].set_title(f"Unsharp Masking (alpha={alpha})")
axes[2, 0].axis('off')

axes[2, 1].imshow(sharpened_spectrum, cmap='gray')
axes[2, 1].set_title("Sharpened Spectrum (High Freq Boost)")
axes[2, 1].axis('off')

# OBSERVATION COMMENT FOR SHARPENING:
# Look at the outer areas of the spectrum above (Row 3, Right).
# NOTICE: The outer areas (High Frequencies) are BRIGHTER/whiter compared to the Original.
# EXPLANATION: Sharpening works by enhancing edges. Edges correspond to high frequency
# changes in pixel intensity. By adding the mask back to the image, we amplified
# the magnitude of these high frequencies, making them brighter in the spectrum.

plt.tight_layout()
plt.show()
````
